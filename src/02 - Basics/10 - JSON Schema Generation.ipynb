## JSON Schema Generation

Wie lässt sich die Struktur eines Models so beschreiben, dass auch andere Systeme sie lesen können, etwa eine API-Dokumentation oder ein Formular-Generator? Dafür gibt es JSON Schema, ein standardisiertes Format zur Beschreibung von JSON-Daten. Pydantic erzeugt es aus den Annotationen eines Models mit `model_json_schema()`.

**Danach ist klar:**
- wie `model_json_schema()` das Schema eines Models liefert,
- wie Typ, Default und Nullable-Eigenschaft im Schema auftauchen,
- woran sich Pflicht-Fields im Schema erkennen lassen.

Voraussetzung: Notebook `09 - Inspecting Fields`.

```python
# `BaseModel` ist die Basisklasse, aus der Pydantic das Schema ableitet.
from pydantic import BaseModel
```

```python
# `field_1` ist nullable und optional, `field_2` ist ein optionaler `str` mit Default.
class Model(BaseModel):
    field_1: int | None = None
    field_2: str = "Python"
```

**Vorhersage:** Wie beschreibt das Schema `field_1` (nullable) und `field_2`? Und gibt es einen Eintrag `required`?

<details>
<summary>Erklärung anzeigen</summary>

**Was der Code macht:** `field_1` erscheint als `anyOf` aus `integer` und `null`, mit `default: None`. `field_2` ist ein `string` mit dem Default `"Python"`. Einen Eintrag `required` gibt es nicht, weil beide Fields einen Default haben. Er würde nur Pflicht-Fields auflisten.

Kurz gesagt: Das Schema bildet Typ, Nullable-Eigenschaft und Default ab, und `required` verrät, was fehlen darf.
</details>

```python
Model.model_json_schema()
```

Das Ergebnis ist ein gewöhnliches `dict` und lässt sich zum Beispiel mit `json.dumps` als JSON-Text weitergeben. Wie lässt sich die verschachtelte Struktur lesbar ausgeben?

```python
# `pprint` gibt das verschachtelte Dictionary übersichtlich über mehrere Zeilen aus.
from pprint import pprint

pprint(Model.model_json_schema())
```

### Zusammenfassung

- `Model.model_json_schema()` liefert das JSON Schema des Models als `dict`.
- Typen und Defaults aus den Annotationen erscheinen im Schema, `int | None` als `anyOf` aus `integer` und `null`.
- `required` listet nur Pflicht-Fields und fehlt, wenn alle Fields einen Default haben.
- `pprint` macht das verschachtelte Ergebnis lesbar.
