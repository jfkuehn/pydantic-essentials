## Serialization

Wie wird aus einer Model-Instanz wieder ein Dictionary oder JSON-Text, etwa für eine API-Antwort oder eine Datei? Das ist die Umkehrung der Deserialization. Pydantic bietet dafür `model_dump` (Dictionary) und `model_dump_json` (JSON-Text).

**Danach ist klar:**
- wie `model_dump` und `model_dump_json` eine Instanz in Dictionary und JSON-Text umwandeln,
- wie sich JSON-Text mit `indent` lesbar formatieren lässt,
- wie `include` und `exclude` einzelne Fields ein- oder ausschließen.

Voraussetzung: Notebook `03 - Deserialization`.

```python
from pydantic import BaseModel, ValidationError
```

```python
class Person(BaseModel):
    first_name: str
    last_name: str
    age : int
```

```python
galois = Person(first_name="Evariste", last_name="Galois", age=20)
newton = Person(first_name='Isaac', last_name='Newton', age=84)
```

Wie liegen die Daten einer Instanz intern vor, und wie werden sie als Dictionary ausgegeben?

```python
# `__dict__` zeigt die Werte, so wie die Instanz sie intern ablegt. Für die Ausgabe taugt `model_dump()` besser,
# weil es Optionen bietet und verschachtelte Models gleich mit umwandelt.
newton.__dict__
```

```python
# `model_dump()` macht aus den Fields der Instanz ein Dictionary.
galois.model_dump()
```

```python
# Ein Blick auf den Typ bestätigt: Es kommt ein ganz gewöhnliches `dict` heraus.
type(galois.model_dump())
```

Wie entsteht daraus JSON-Text? Ein Umweg über `json.dumps` ist nicht nötig.

```python
# `model_dump_json()` liefert gleich fertigen JSON-Text, kompakt und ohne Einrückung.
newton.model_dump_json()
```

```python
# Das Ergebnis ist ein `str` und kein `dict`, also reiner Text.
type(newton.model_dump_json())
```

```python
# Mit `indent=2` wird der JSON-Text eingerückt. `print` gibt ihn mit echten Zeilenumbrüchen aus,
# sonst stünden lauter `\n` mitten im Text.
print(newton.model_dump_json(indent=2))
```

### Fields ein- und ausschließen

Wie lässt sich nur ein Teil der Daten ausgeben, etwa ohne `age`? Dafür gibt es `exclude` und `include`.

**Vorhersage:** Welche Fields bleiben übrig, wenn `first_name` und `age` ausgeschlossen werden?

<details>
<summary>Erklärung anzeigen</summary>

**Was der Code macht:** `exclude` nimmt eine Liste von Fieldnamen und lässt sie im Ergebnis weg. Übrig bleibt `last_name`.
</details>

```python
galois.model_dump(exclude=["first_name", "age"])
```

```python
# `model_dump_json` kennt dieselben Optionen wie `model_dump`, hier fehlt `age` im JSON-Text.
newton.model_dump_json(exclude=["age"])
```

<details>
<summary>Erklärung anzeigen</summary>

**Was der Code macht:** `include` ist das Gegenstück: Nur die genannten Fields bleiben im Ergebnis.

Kurz gesagt: `exclude` nennt, was fehlt, `include` nennt, was bleibt.
</details>

```python
newton.model_dump(include=["last_name"])
```

### Zusammenfassung

- `model_dump()` liefert ein Dictionary, `model_dump_json()` JSON-Text (`str`).
- `indent` formatiert den JSON-Text lesbar.
- `exclude` und `include` schließen Fields aus oder beschränken die Ausgabe, in beiden Methoden.
- `__dict__` zeigt die intern gespeicherten Werte, `model_dump()` ist der vorgesehene Weg.
